```python
import pandas as pd
import numpy as np

# Step 1: Load original CSV for calculation
file_path = 'interop_orig.csv'
df_calc = pd.read_csv(file_path)

# Step 2: Check if required columns exist
required_cols = ['TimeStamp', 'longAccel', 'latAccel', 'vertAccel']
if not all(col in df_calc.columns for col in required_cols):
    raise ValueError(f"Missing required columns in CSV: {required_cols}")

# Step 3: Convert TimeStamp to seconds
df_calc['TimeStamp'] = pd.to_timedelta(df_calc['TimeStamp']).dt.total_seconds()

# Step 4: Compute resultant acceleration
df_calc['resultant_acc'] = np.sqrt(
    df_calc['latAccel']**2 +
    df_calc['longAccel']**2 +
    df_calc['vertAccel']**2
)

# Step 5: Sort and compute delta_t
df_calc = df_calc.sort_values('TimeStamp')
df_calc['delta_t'] = df_calc['TimeStamp'].diff().fillna(0)

# Step 6: Compute velocity
df_calc['velocity'] = (df_calc['resultant_acc'] * df_calc['delta_t']).cumsum()

# Step 7: Extract only new columns
df_new = df_calc[['resultant_acc', 'velocity']]

# Step 8: Load target CSV and append new columns
new_file_path = 'harman_clean.csv'
df_target = pd.read_csv(new_file_path)

# Ensure same number of rows
if len(df_target) != len(df_new):
    raise ValueError("Row mismatch: target CSV and computed data must have same number of rows.")

# Add new columns to existing target data
df_target = df_target.copy()
df_target['resultant_acc'] = df_new['resultant_acc'].values
df_target['velocity'] = df_new['velocity'].values

# Save updated CSV
df_target.to_csv(new_file_path, index=False)

print(f"File '{new_file_path}' updated with only 'resultant_acc' and 'velocity' columns.")
```

```python
from datetime import datetime
import pandas as pd
from pydantic import BaseModel


class Point(BaseModel):
    latitude: float
    longitude: float
    timestamp: float  # Unix timestamp
    yaw_rate: float = 0.0 
    velocity: float = 0.0   # Optional velocity
    resultant_acc: float = 0.0  # Optional resultant acceleration

def extract_points_from_csv(
    file_path: str, 
    lat_col: str = "lat", 
    lon_col: str = "long", 
    time_col: str = "TimeStamp", 
    yaw_col: str = "yawRate",
    velocity_col: str = "velocity",
    acc_col: str = "resultant_acc"
) -> list[Point]:

    df = pd.read_csv(file_path)
    points = []

    # Use today's date or a fixed base date
    base_date = datetime(2025, 2, 21).date()
    
    for _, row in df.iterrows():
        try:
            # Combine date and time into a full datetime
            full_datetime_str = f"{base_date} {row[time_col]}"
            full_datetime = datetime.strptime(full_datetime_str, "%Y-%m-%d %H:%M:%S.%f")
            timestamp = full_datetime.timestamp()

            # Read values or default
            lat = float(row[lat_col])
            long = float(row[lon_col])
            yaw = float(row[yaw_col]) if yaw_col in row else 0.0
            vel = float(row[velocity_col]) if velocity_col in row else 0.0
            acc = float(row[acc_col]) if acc_col in row else 0.0

            # Create the Point object
            point = Point(
                latitude=lat,
                longitude=long,
                timestamp=timestamp,
                yaw_rate=yaw,
                velocity=vel,
                resultant_acc=acc
            )
            points.append(point)
        except (ValueError, TypeError, KeyError):
            continue  # Skip bad rows

    return points

# Example usage
points = extract_points_from_csv("harman_clean.csv")
print(len(points))
```

```python
import requests
import folium

osrm_matched_points = []
fmap = folium.Map(location=[points[0].latitude, points[0].longitude], zoom_start=16)
BATCH_SIZE = 100

print(f"Total batches expected: {(len(points) + BATCH_SIZE - 1) // BATCH_SIZE}")

for start in range(0, len(points), BATCH_SIZE):
    batch = points[start:start + BATCH_SIZE]

    if len(batch) < 2:
        continue  # OSRM match requires at least 2 points
    try:
        coordinates_str = ';'.join([f"{point.longitude},{point.latitude}" for point in batch])
        radiuses_str = ';'.join(['50'] * len(batch))
        yaw_rates_str = ';'.join([str(point.yaw_rate) for point in batch])
    except Exception as e:
        print(f"Error generating batch {start // BATCH_SIZE + 1}: {e}")
        continue
    osrm_match_url = (
        f"http://127.0.0.1:5001/match/v1/driving/{coordinates_str}"
        f"?overview=full&geometries=geojson&gaps=ignore&radiuses={radiuses_str}&yaw_rate={yaw_rates_str}"
    )

    response = requests.get(osrm_match_url)
    print(f"\nBatch {start // BATCH_SIZE + 1}: Requested URL:")
    print(osrm_match_url)
    print("Status Code:", response.status_code)

    if response.status_code != 200:
        print("Request failed. Skipping batch.")
        continue

    data = response.json()

    if data["code"] != "Ok" or "matchings" not in data or not data["matchings"]:
        print("Matching failed or no matchings returned.")
        continue

    # Plot matched coordinates (as line)
    for matching in data["matchings"]:
        matched_coords = matching["geometry"]["coordinates"]
        matched_coords = [[lat, lon] for lon, lat in matched_coords]  # Flip for folium
        folium.PolyLine(
            locations=matched_coords,
            color='green',
            weight=3,
        ).add_to(fmap)

    # Plot tracepoints (snapped points)
    for i, tp in enumerate(data.get("tracepoints", [])):
        if tp and tp.get("location"):
            lon, lat = tp["location"]
            
            orig_point = batch[i]
        # Create new Point with matched lat/lon but other original attributes
            matched_point = Point(
                latitude=lat,
                longitude=lon,
                timestamp=orig_point.timestamp,
                yaw_rate=orig_point.yaw_rate,
                velocity=orig_point.velocity,
                resultant_acc=orig_point.resultant_acc
            )

            osrm_matched_points.append(matched_point)
            
            folium.CircleMarker(
                location=[lat, lon],
                radius=2,
                color='blue',
                fill=True,
                popup=f"Snapped Point {start + i}"
            ).add_to(fmap)
    
    for p in batch:
        folium.CircleMarker(
            location=[p.latitude, p.longitude],
            radius=2,
            color='purple',
            fill=True,
            fill_color='purple'
        ).add_to(fmap)
        
    folium.PolyLine(
            locations=[(p.latitude, p.longitude) for p in batch],
            color='red',
            weight=3,
            opacity=.7
        ).add_to(fmap)
    
            
print("Total points:", len(points))
print("Any point without yaw_rate?", any(not hasattr(p, "yaw_rate") for p in points))
fmap
```

```python
from geopy.distance import distance
from geopy.point import Point as GeoPoint
import math

def compute_lateral_offset(yaw_deg: float, velocity: float, delta_t: float = 100.0) -> float:
    if abs(yaw_deg) < 1e-5 or abs(velocity) < 1e-5:
        return 0.0
    omega = math.radians(yaw_deg)
    radius = velocity / omega
    return radius * (1 - math.cos(omega * delta_t))

def heading_from_points(p1: Point, p2: Point) -> float:
    dx = p2.longitude - p1.longitude
    dy = p2.latitude - p1.latitude
    return math.atan2(dy, dx)  # in radians

def offset_latlon(lat, lon, dx_meters, dy_meters):
    origin = GeoPoint(lat, lon)
    shifted_north = distance(meters=dy_meters).destination(origin, 0)
    shifted_east = distance(meters=dx_meters).destination(shifted_north, 90)
    return shifted_east.latitude, shifted_east.longitude

def compute_adjusted_latlon_from_osrm(points: list[Point]) -> list[tuple[float, float]]:
    adjusted_coords = []
    for i in range(1, len(points)):
        p1 = points[i - 1]
        p2 = points[i]

        # Compute heading angle (in radians)
        heading = heading_from_points(p1, p2)

        # Time delta (use timestamp)
        delta_t = p2.timestamp - p1.timestamp
        if delta_t <= 0:
            delta_t = 1  # fallback to 1s

        # Lateral offset using yaw rate and velocity
        lateral_offset = compute_lateral_offset(
            yaw_deg=p2.yaw_rate,
            velocity=p2.velocity,
            delta_t=delta_t
        )

        # Left normal vector
        dx = lateral_offset * -math.sin(heading)
        dy = lateral_offset *  math.cos(heading)

        # Apply offset to current point
        adj_lat, adj_lon = offset_latlon(p2.latitude, p2.longitude, dx, dy)
        adjusted_coords.append((adj_lat, adj_lon))
    return adjusted_coords
```

```python
adjusted_coords = compute_adjusted_latlon_from_osrm(osrm_matched_points)

# Initialize map at first OSRM-matched point
fmap = folium.Map(location=[osrm_matched_points[0].latitude, osrm_matched_points[0].longitude], zoom_start=16)

# Plot matched OSRM points (blue)
for p in osrm_matched_points:
    folium.CircleMarker(
        location=[p.latitude, p.longitude],
        radius=4,
        color='blue',
        fill=True,
        popup='Matched'
    ).add_to(fmap)

# Plot adjusted lateral-shifted points (green)
for lat, lon in adjusted_coords:
    folium.CircleMarker(
        location=[lat, lon],
        radius=2,
        color='green',
        fill=True,
        popup='Adjusted'
    ).add_to(fmap)

# Optional: Connect adjusted points with a green polyline
folium.PolyLine(locations=adjusted_coords, color='green', weight=2, opacity=0.7).add_to(fmap)

# Show map
fmap
```
